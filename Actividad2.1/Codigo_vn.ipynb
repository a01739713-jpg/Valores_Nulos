Cuarto metodo de sustitución: Sustituir valores nulos por un string en concreto 

```python
#Me apoye con IA para generar esta parte del código para que rellenara las fechas vacias con el mes y año correspondiente
rangos_meses = [
    ('Abril', '2025', 0, 30), ('Mayo', '2025', 34, 58), ('Junio', '2025', 61, 86), ('Julio', '2025', 88, 120),
    ('Agosto', '2025', 122, 166), ('Septiembre', '2025', 168, 230), ('Octubre', '2025', 232, 290),
    ('Noviembre', '2025', 292, 343), ('Diciembre', '2025', 345, 391), ('Enero', '2026', 393, 428),
    ('Febrero', '2026', 430, 460), ('Marzo', '2026', 462, 477), ('Abril', '2026', 479, 533),
    ('Mayo', '2026', 540, 598), ('Junio', '2026', 600, 640), ('Julio', '2026', 642, 675),
]

for mes, anio, ini, fin in rangos_meses:
    data.loc[ini:fin, 'Fecha'] = data.loc[ini:fin, 'Fecha'].fillna(f'{mes} {anio}')
```
