```python
data = pd.read_csv('Citas_Digital_Filtrado.csv')
#data.head()
```

```python
#Identificar valores nulos por columna
valores_nulos=data.isnull().sum()
valores_nulos
```

Eliminar columnas innecesarias

```python
# Se decidió eliminar la columna Telefono porque no tenia ningun registro
data2 = data.drop(columns=['Telefono '])

# Verificar resultado
print(data2.head())
```

Cuarto metodo de sustitución: Sustituir valores nulos por un string en concreto 

```python
#Me apoye con IA para generar esta parte del código para que rellenara las fechas vacias con el mes y año correspondiente
rangos_meses = [
    ('Abril', '2025', 0, 30), ('Mayo', '2025', 34, 58), ('Junio', '2025', 61, 86), ('Julio', '2025', 88, 120),
    ('Agosto', '2025', 122, 166), ('Septiembre', '2025', 168, 230), ('Octubre', '2025', 232, 290),
    ('Noviembre', '2025', 292, 343), ('Diciembre', '2025', 345, 391), ('Enero', '2026', 393, 428),
    ('Febrero', '2026', 430, 460), ('Marzo', '2026', 462, 477), ('Abril', '2026', 479, 533),
    ('Mayo', '2026', 540, 598), ('Junio', '2026', 600, 640), ('Julio', '2026', 642, 675),
]

for mes, anio, ini, fin in rangos_meses:
    data2.loc[ini:fin, 'Fecha'] = data2.loc[ini:fin, 'Fecha'].fillna(f'{mes} {anio}')
```

```python
data2.head()
```

Eliminar las filas de nombre completo

```python
#Se eligió nombre completo porque es la columna que considero más importante para considerar si se borra 
#la columna completa o no
data2 = data2.dropna(subset=['Nombre Cliente '])
```
