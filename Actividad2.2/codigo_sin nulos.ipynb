```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Análisis_de_piso.xlsx')
df.head()
```

```python
df_sinnulos = df.isnull().sum()
df_sinnulos
```

```python
# Seleccionamos las columnas de meses (leads 2024, leads 2025, ventas 2024, ventas 2025)
columnas_meses = df.columns[1:16].tolist() + df.columns[20:37].tolist()

# Rellenamos esas columnas con 0
df[columnas_meses] = df[columnas_meses].fillna(0)
```
